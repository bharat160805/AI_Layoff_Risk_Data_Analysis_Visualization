**Detect Outliers:**
- With the help of IQR method we find the lower_bound and upper_bound and outliers.
- Calculate the number of outliers in each columns then print them.
- Use the boxplot to show the outliers easily on the graph.

```python
numeric_columns=df.select_dtypes(include='number').columns
for col in numeric_columns:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    print("\nColumn:",col)
    print(f"Lower bound of {col}:",lower_bound)
    print(f"Upper bound of {col}:",upper_bound)
    outliers=df[(df[col]<lower_bound)|(df[col]>upper_bound)]
    print("\nOutliers:",outliers)
    print("\nNumber of outliers:",len(outliers))
```

```python
# Boxplot to represent the outliers
df[numeric_columns].boxplot(figsize=(14,10))
plt.xticks(rotation=60)
plt.show()
```

**Remove the outliers:**

```python
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound)&(df[col] <= upper_bound)]
```

```python
#The boxplot after removing the out liers bu IQR method: 
df[numeric_columns].boxplot(figsize=(15,8))
plt.xticks(rotation=45)
plt.show()
```
